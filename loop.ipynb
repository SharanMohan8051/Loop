```python
import json

with open("university.json", "r") as file:
    data = json.load(file)

uni = data["university"]
print(uni["name"])
```

Nested Loop
1. Faculty and department pairs (2 levels). Print every "Faculty → Department" pair.

```python
for i in data["university"]["faculties"]:
  for j in i["departments"]:
    print(j["name"],j["head"])
```

2. All courses (3 levels: faculties, departments, courses). Print every course code, title and credits. Then print the total credits offered.

```python
for i in data["university"]["faculties"]:
  for j in i["departments"]:
    for z in j["courses"]:
      print(z['code'],z['title'],z['credits'])
```

3. All students with their course (4 levels). Print "<student name> – <course code>" for every enrolled student in the university.

```python
for i in data["university"]["faculties"]:
  for j in i["departments"]:
    for z in j["courses"]:
      for k in z['enrolled_students']:
        print(z["code"],k["name"])
```

4. Pending work report (5 levels, down to assignments). List every pending assignment along with the student’s name and due date.

```python
for i in data["university"]["faculties"]:
  for j in i["departments"]:
    for z in j["courses"]:
      for k in z['enrolled_students']:
        for l in k['assignments']:
          print(k['name'],l['name'])
```
